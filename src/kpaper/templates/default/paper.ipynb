```python
from kbasic.parsing import TOML, File, Folder
from importlib.metadata import version 
class Author:
    def __init__(
            self, 
            name: str, 
            affiliation: str = None, 
            corresponding: str = None, 
            orcid: str = None
        ) -> None:
        self.name = name 
        self.affiliation = affiliation
        self.corresponding = corresponding
        self.orcid = orcid

class KPaperConfig(TOML):
    def __init__(
            self, 
            parent: Folder,
            title: str = r"''",
            shorttitle: str = r"''",
            shortauthors: str = r"''",
            keywords: list[str] = [],
            abstract: str = r"''",
            authors: dict[str, list] = {
                "names":["Keyan Gootkin",],
                "orcids":["0000-0003-0922-138X",],
                "emails":["gootkin@hawaii.edu",],
                "affiliations":["Institute for Astronomy, University of Hawaii, Manoa, 2680 Woodlawn Dr., Honolulu, HI 96822, USA",],
                "corresponding":0
            },
            editorial: dict[str, str] = {
                "received": r"''",
                "revised": r"''",
                "accepted": r"''",
                "published": r"''",
                "submitjournal": r"''"
            },
            document_class: list[str] = ['twocolumn','aastex7'],
            bibliography: str = "",
            load = False
            ) -> None:
        self.parent: Folder = parent
        File.__init__(self, parent.path+'/kpaper.toml')
        self.document_class = document_class
        self.title = title 
        self.shorttitle = shorttitle 
        self.authors = authors
        self.shortauthors = shortauthors
        self.abstract = abstract 
        self.keywords = keywords 
        self.editorial = editorial
        self.bibliography = bibliography
        
        if load: self.read()

    @property 
    def lines(self) -> list[str]:
        return [
            f"# KPaper v.{version('kpaper')} Configuration File",
            f"document_class = {self.document_class}",
            f"title = {self.title}",
            f"shorttitle = {self.shorttitle}",
            f"shortauthors = {self.shortauthors}",
            f"abstract = {self.abstract}",
            f"keywords = {self.keywords}",
            f"bibliography = {self.bibliography}",
            "",
            "[authors]",
            f"names = {self.authors['names']}",
            f"orcids = {self.authors['orcids']}",
            f"emails = {self.authors['emails']}",
            f"affiliations = {self.authors['affiliations']}",
            f"corresponding = {self.authors['corresponding']}",
            "",
            f"[editorial]",
            f"received = {self.editorial['received']}",
            f"revised = {self.editorial['revised']}",
            f"accepted = {self.editorial['accepted']}",
            f"published = {self.editorial['published']}",
            f"submitjournal = {self.editorial['submitjournal']}",
        ]
    
    def make_tex(self):
        f = File(f'{self.parent.path}/LaTeX/main.tex')
        author_segments = "\n".join("\n".join([
                r"\author["+str(o)+"]{"+n+"}",
                r"\email{"+e+"}",
                r"\affiliation{"+a+"}"])
             for n,o,e,a in zip(self.authors["names"], self.authors["orcids"], self.authors["emails"], self.authors["affiliations"])
        )
        print(author_segments)
        f.lines = [
            r"\documentclass[" + str(self.document_class[:-1]) + "]{" + self.document_class[-1] + r"}",
            r"\usepackage[utf8]{inputenc}",
            r"\begin{document}",
            r"\title{"+self.title+"}",
            r"\shorttitle{"+self.shorttitle+"}",
        ] + author_segments.split("\n") + [
            r"\corresponding_author{" + str(self.authors['names'][self.authors['corresponding']]) + "}",
            r"\begin{abstract}",
            self.abstract,
            r"\end{abstract}",
            r"\keywords{"+str(self.keywords)+"}",
            r"",
            r"",
            r"",
            r"\end{document}"
        ]
        return f
```

```python
p = Folder("./")
c = KPaperConfig(p)
```

```python
f = c.make_tex()
f.save()
```
